# CSV shelf life quick check

This notebook can be used to sanity-check how the backend CSV ingestion will interpret shelf-life columns.

- Boolean `More than 1 Day Shelf Life` column: TRUE/Yes/Y/1 → shelf_life_days = 2, anything else non-empty → 1.
- Numeric `shelf_life` column: values >1 → 2, values <=1 → 1 (clamped to 1–2).

Use a small sample CSV to confirm inference and ingestion end-to-end when needed.
